# **Analisis de Series de Tiempo II**

# **Clase 5, Convoluciones 1D**

Importamos lo necesario

```python
import numpy as np
import tensorflow as tf
```

Definimos señal de entrada

```python
x = np.array([[1, 2, 3, 4, 5]], dtype=np.float32)
print(x)
```

Convertimos a entrada que recibe TF

```python
# (batch, longitud, canales)
x = x.reshape(1, 5, 1)
```

## **A) Conv1D Normal**

Definimos convolución normal

```python
conv = tf.keras.layers.Conv1D(
    filters=1,
    kernel_size=3,
    padding="same",
    use_bias=False
)
```

Construimos la capa

```python
conv.build((1,5,1))
```

Definimos manualmente el kernel

```python
kernel = np.array([[[1]],
                   [[1]],
                   [[1]]], dtype=np.float32)

conv.set_weights([kernel])
```

Resultado

```python
y = conv(x)
```

```python
print("Entrada:")
print(x[0,:,0])

print("\nSalida:")
print(y.numpy()[0,:,0])
```

## **B) Conv1D Causal**

Definimos convolución causal

```python
conv = tf.keras.layers.Conv1D(
    filters=1,
    kernel_size=3,
    padding="causal",
    use_bias=False
)
```

```python
conv.build((1,5,1))

kernel = np.array([[[1]],
                   [[1]],
                   [[1]]], dtype=np.float32)

conv.set_weights([kernel])
```

Resultado

```python
y = conv(x)

print("Entrada")
print(x[0,:,0])

print("\nSalida")
print(y.numpy()[0,:,0])
```

## **C) Conv1D Dilatada**

```python
conv = tf.keras.layers.Conv1D(
    filters=1,
    kernel_size=3,
    dilation_rate=2,
    padding="same",
    use_bias=False
)
```

```python
conv.build((1,8,1))

kernel = np.array([[[1]],
                   [[1]],
                   [[1]]], dtype=np.float32)

conv.set_weights([kernel])
```

Resultado

```python
y = conv(x)

print("Entrada")
print(x[0,:,0])

print("\nSalida")
print(y.numpy()[0,:,0])
```
